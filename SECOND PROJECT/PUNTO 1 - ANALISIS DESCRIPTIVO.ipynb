# Imports Necesarios

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec
```

# Carga del Dataset

```python
df = pd.read_csv('avocado.csv')
df.head()
```

## ANALISIS EXPLORATORIO

```python
df.info()
```

### *Correlación por medio de heatmaps*

```python
correlation = df.select_dtypes(include='number').corr()
plt.figure(figsize=(10,6))
sns.heatmap(correlation, annot=True, cmap="coolwarm")
plt.show()
```

### *Histograma*

```python
num_cols = ["AveragePrice", "Total Volume", "4046", "4225", "4770",
             "Total Bags", "Small Bags", "Large Bags", "XLarge Bags"]

fig, axes = plt.subplots(nrows=3, ncols=3, figsize=(14, 10))
axes = axes.flatten()

for i, col in enumerate(num_cols):
    axes[i].hist(df[col].dropna(), bins=40, color="#7F77DD", edgecolor="white", alpha=0.85)
    axes[i].set_title(col, fontsize=11)
    axes[i].set_xlabel("")

plt.suptitle("Distribución de variables numéricas", fontsize=14, y=1.01)
plt.tight_layout()
plt.show()
```

### *Boxplots*

```python
fig, ax = plt.subplots(figsize=(8, 5))
sns.boxplot(
    data=df,
    x="type",
    y="AveragePrice",
    palette="Set2",
    ax=ax
)
ax.set_title("Precio por tipo")
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 6))
sns.boxplot(
    data=df,
    x="type",
    y="Total Volume",
    palette="Set2",
    ax=ax,
    showfliers=False
)
ax.set_title("Volumen Total por Tipo")
ax.set_ylabel("Volumen Total")
plt.tight_layout()
plt.show()
```

### *Grafico de barras*

```python
fig, ax = plt.subplots(figsize=(16, 6))
sns.barplot(
    data=df,
    x="region",
    y="Total Volume",
    hue="type",
    palette="Set2",
    ax=ax
)
ax.set_title("Volumen de ventas por región y tipo de aguacate")
ax.set_xlabel("Región")
ax.set_ylabel("Volumen promedio vendido")
plt.xticks(rotation=45, ha="right")  
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(16, 6))
sns.barplot(
    data=df,
    x="year",
    y="Total Bags",
    palette="Set2",
    ax=ax
)
ax.set_title("Total Bags vendido por año")
ax.set_xlabel("Año")
ax.set_ylabel("Total Bags vendido")
plt.xticks(rotation=45, ha="right")  
plt.tight_layout()
plt.show()
```
